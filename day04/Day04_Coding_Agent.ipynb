```python
from typing import TypedDict


class CodingState(TypedDict):

    requirement:str

    files:list

    code:str

    review:str
```

```python
def project_analyzer(state):

    files = read_project_files(
        "./sample_project"
    )


    state["files"] = files


    return state
```

```python
def coder_agent(state):


    requirement = state["requirement"]


    code=f"""

// 自动生成代码

//需求:
// {requirement}


using UnityEngine;


public class ObjectPool
{{


}}


"""


    state["code"] = code


    return state
```

```python
def reviewer_agent(state):


    code = state["code"]


    if "class" in code:

        state["review"] = "代码结构正常"

    else:

        state["review"] = "代码错误"


    return state
```

```python
from langgraph.graph import StateGraph


workflow = StateGraph(
    CodingState
)


workflow.add_node(
    "Analyzer",
    project_analyzer
)


workflow.add_node(
    "Coder",
    coder_agent
)


workflow.add_node(
    "Reviewer",
    reviewer_agent
)



workflow.set_entry_point(
    "Analyzer"
)


workflow.add_edge(
    "Analyzer",
    "Coder"
)


workflow.add_edge(
    "Coder",
    "Reviewer"
)


app = workflow.compile()
```

```python
result = app.invoke(
{
    "requirement":
    "创建Unity对象池系统"
}
)


print(result)
```
