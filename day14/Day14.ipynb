# Day14 Agent Evaluation

## Goal
Build a deterministic offline benchmark for coding-agent quality gates and operational metrics.

## Contract and metrics

The versioned fixture measures compile success, repair success, token availability, loop counts, and functional stability.

```python
from pathlib import Path
from evaluation.schema import load_suite, load_acceptance
from evaluation.metrics import evaluate_suite
from evaluation.report import render_json, render_markdown
root = Path.cwd()
suite = load_suite(root / 'evaluation/cases/day14_benchmark.json')
acceptance = load_acceptance(root / 'evaluation/integration/day14_real_acceptance.json')
result = evaluate_suite(suite)
```

## Deterministic report

Headline metrics are pinned and rendering the same inputs twice must produce identical bytes.

```python
headline = {
    'end_to_end': result.end_to_end_success.fraction,
    'compile': result.compile_success.fraction,
    'repair': result.repair_success.fraction,
    'stability': result.functional_stability.fraction,
}
print(headline)
assert headline == {'end_to_end': (2, 4), 'compile': (2, 3), 'repair': (1, 2), 'stability': (5, 5)}
assert render_json(result, acceptance) == render_json(result, acceptance)
assert render_markdown(result, acceptance) == render_markdown(result, acceptance)
```

## Real acceptance separation

Real integration evidence is loaded separately and never changes the offline benchmark score. Missing scenarios remain not recorded.

```python
print({'recorded_real_scenarios': len(acceptance.scenarios), 'offline_cases': len(result.cases)})
assert len(result.cases) == 5
```
