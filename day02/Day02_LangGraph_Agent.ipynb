```python
# 环境初始化
from typing import TypedDict

from langgraph.graph import (
    StateGraph,
    END
)

print("LangGraph环境正常")
```

```python
# 定义状态
# 这里是整个Agent系统的数据中心。
class AgentState(TypedDict):

    # 用户任务
    task: str

    # 需求分析结果
    requirements: str

    # 架构设计结果
    architecture: str

    # 代码结果
    code: str
```

```python
graph = StateGraph(
    AgentState
)


print("Graph创建成功")
```

```python
# 执行需求分析Agent
def requirement_agent(state):

    print("执行需求分析Agent")


    state["requirements"] = """

需求分析:

1. 玩家可以添加物品
2. 玩家可以删除物品
3. 支持背包容量限制
4. UI显示物品


"""


    return state
```

```python
test_state={

"task":
"开发Unity背包系统",

"requirements":
"",

"architecture":
"",

"code":
""

}


requirement_agent(test_state)
```

```python
# 执行架构设计Agent
def architecture_agent(state):


    print(
        "执行架构设计Agent"
    )


    state["architecture"]="""

系统架构:

InventoryManager
负责背包逻辑

ItemData
负责物品数据

BagUI
负责界面显示

"""


    return state
```

```python
# 添加Node
graph.add_node(
    "requirement",
    requirement_agent
)


graph.add_node(
    "architecture",
    architecture_agent
)


print("Node添加完成")
```

```python
# 连接流程
graph.set_entry_point(
    "requirement"
)



graph.add_edge(
    "requirement",
    "architecture"
)



graph.add_edge(
    "architecture",
    END
)


print("流程连接完成")
```

```python
# 编译
app = graph.compile()

print("Agent Workflow编译完成")
```

```python
# 运行
result = app.invoke({

"task":
"开发Unity背包系统",

"requirements":
"",

"architecture":
"",

"code":
""

})


result
```
